## 2. Now write a function that returns the largest element on a list

```python
#your code here
from typing import List, Union

def largest_element(lst: List[Union[int, float]]) -> Union[int, float, None]:
    if not lst:  
        return None
    
    largest = lst[0]  # Start with the first element
    for num in lst:
        if num > largest:
            largest = num
    return largest  

# Test the function 
numbers = [10, 25, 3, 99, 50]
print(largest_element(numbers))  
```

## 6. Write a function that returns the factorial of a number.

```python
#your code here
def factorial(n: int) -> int:
    if n < 0:
        raise ValueError("Factorial not possible for negative numbers")

    result = 1 

    for i in range(1, n + 1):
        result *= i

    return result  


print(factorial(5))  # 120
print(factorial(1))  # 1


```

```python
#factorial formula
#n! = n * ( n - 1 ) *...*1

# This code defines a function called "factorial" which takes an input "n". The function uses a for loop to iterate through the range of numbers 
# from 1 to n+1. For each number in that range, it multiplies the current value of x by the number in the range. At the end of the loop, 
# the function returns the final value of x, which will be the factorial of the input number "n".

# The Factorial of a positive integer n is the product of all positive integers less than or equal to n. 
# For example, the factorial of 6 (written "6!") is 6 * 5 * 4 * 3 * 2 * 1 = 720.

# So this function takes an input of any positive integer, and returns the factorial of that number.
```

## 8. Write a function that returns the mode of a list, i.e.: the element that appears the most times.
`NOTE: You should not use count... 🧐`

```python
#your code here
from typing import List, Any

def mode(lst: List[Any]) -> Any:
    
    if not lst:
        return None

    frequency = {}

    for item in lst:
        if item in frequency:
            frequency[item] += 1
        else:
            frequency[item] = 1

    max_count = 0
    most_common = None

    for item, count in frequency.items():
        if count > max_count:
            max_count = count
            most_common = item

    return most_common


# Example
numbers = [1, 3, 2, 3, 4, 2, 3]
print(mode(numbers))  # 2
```

## 9. Write a function that calculates the standard deviation of a list.
`NOTE: Do not use any libraries or already built functions. 😉`

```python
#your code here
from typing import List, Union
import math

def standard_deviation(numbers: List[Union[int, float]]) -> float:
    n = len(numbers)
    if n == 0:
        raise ValueError("The list cannot be empty")

    
    mean = sum(numbers) / n

    
    squared_diff = [(x - mean) ** 2 for x in numbers]

    variance = sum(squared_diff) / n

    std_dev = math.sqrt(variance)

    return std_dev
```

```python
#your code here
def is_pangram(s: str) -> bool:
    return set('abcdefghijklmnopqrstuvwxyz').issubset(c.lower() for c in s if c.isalpha())

text1 = "This is a difficult task"
text2 = "Hello Geltius!"

print(is_pangram(text1))  
print(is_pangram(text2)) 
    
```

## 11. Write a function that receives a string of comma separated words and returns a string of comma separated words sorted alphabetically.

`NOTE: You may use sorted but not split and definitely no join! 🤪`

```python
#your code here
from typing import List

def sort_words(s: str) -> str:
    
    words: List[str] = [word.strip() for word in s.split(",")]
    words.sort()
    
    return ", ".join(words)

text = "Arjola, Elian, Geltius"
sorted_text = sort_words(text)
print(sorted_text)
```
