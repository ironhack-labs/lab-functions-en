## 6. Write a function that returns the factorial of a number.

```python

#your code here
 
def factorial(n):
    if n < 0:
        raise ValueError("El factorial no esta definido para numeros negativos")
    result = 1
    for i in range (1, n+1):
        result = result * i
    return result
num = int(input("Ingrese el numero para calcular el factorial:"))
print(f"\nEl factorial de {num} es:", factorial(num))
```

```python
#factorial formula
#n! = n * ( n - 1 ) *...*1

# This code defines a function called "factorial" which takes an input "n". The function uses a for loop to iterate through the range of numbers 
# from 1 to n+1. For each number in that range, it multiplies the current value of x by the number in the range. At the end of the loop, 
# the function returns the final value of x, which will be the factorial of the input number "n".

# The Factorial of a positive integer n is the product of all positive integers less than or equal to n. 
# For example, the factorial of 6 (written "6!") is 6 * 5 * 4 * 3 * 2 * 1 = 720.

# So this function takes an input of any positive integer, and returns the factorial of that number.
```

## 7. Write a function that takes a list and returns a list of the unique values.

`NOTE: You cannot use set. 🤔`

```python

#your code here
def unique(lst_un):
    unique_values = []
    for i in lst_un:
        if i not in unique_values:
            unique_values.append(i)
    return unique_values

lst = [5,5,5,6,7,["alpha","betha","gamma"],8,9,9,9,5,1,2,100,["alpha","betha","gamma"],45,8,2,9,50,41,45]

print("La lista original es:", lst)
print("La lista con valores unicos es:",unique(lst))
```

```python
# This will test your function 
test_unique(unique)
```

## 8. Write a function that returns the mode of a list, i.e.: the element that appears the most times.
`NOTE: You should not use count... 🧐`

```python

#your code here
def mode_counter(lst):
    frecuencia = {}
    for i in lst:
        if i in frecuencia:
            frecuencia[i] += 1
        else: 
            frecuencia[i] = 1
    max_frec = 0
    moda_valor = None
    for key, valor in frecuencia.items():
        if valor > max_frec:
            max_frec = valor
            moda_valor = key
            
    return moda_valor

mi_lista = [5, 3, 9, 5, 2, 5, 3, 3, 3, 7]
print("La lista es:", mi_lista)
print("La moda de la lista es:", mode_counter(mi_lista))
```

## 9. Write a function that calculates the standard deviation of a list.
`NOTE: Do not use any libraries or already built functions. 😉`

```python

#your code here
def st_dev(list_sd):
    n = len(list_sd)
    if n == 0: 
        return 0 
    mean = sum(list_sd)/n
    sum_diff_sq = 0
    for x in list_sd:
        sum_diff_sq += (x - mean)**2
    return (sum_diff_sq/n)**0.5

lista = [2, 4, 4, 4, 5, 5, 7, 9]
print("Lista:", lista)
print("Desviación estándar:", st_dev(lista))
```

```python
# This will test your function 
test_stdev(st_dev)
```

## 11. Write a function that receives a string of comma separated words and returns a string of comma separated words sorted alphabetically.

`NOTE: You may use sorted but not split and definitely no join! 🤪`

```python
#your code here

def sort_alpha(string):
    palabras = []
    palabra = ""
    
    for i in string:
        if i == ",":
            palabras.append(palabra)
            palabra =""
        else:
            palabra = palabra + i
            
    palabras.append(palabra)
    
    palabras_ordenadas = sorted(palabras)
    
    result = ""
    
    for i, w in enumerate(palabras_ordenadas):
        result = result + w
        if i != len(palabras_ordenadas) - 1:
            result = result + "," 
    return result  

ejemplo = "banana,apple,cherry,date,murcielago,zorrillo,racoon"
print("Original:", ejemplo)
print("Ordenado:", sort_alpha(ejemplo))
  
            
```

## 12. Write a function to check if a given password is strong (at least 8 characters, at least one lower case, at least one upper case, at least one number and at least one special character). It should output True if strong and False if not.

```python

#your code here
def check_pass(password):
    if len(password) < 8:
        return False
    minusculas = False
    mayusculas = False
    numeros = False
    especiales = False

    digitos_especiales = "!@#$%^&*()-_+=[]{}|:;,.<>?/`~"
    
    for i in password:
        if i.islower():
            minusculas = True
        elif i.isupper():
            mayusculas = True
        elif i.isdigit():
            numeros = True
        elif i in digitos_especiales:
            especiales = True
    return minusculas and mayusculas and numeros and especiales

print(check_pass("Abc123!@"))   
print(check_pass("abcdefg"))    
print(check_pass("ABC12345"))   
print(check_pass("abcABC123")) 
```

```python
# This will test your function 
test_pass(check_pass)
```
