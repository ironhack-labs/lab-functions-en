## 6. Write a function that returns the factorial of a number.

```python
def factorial (n):
    fact = 1
    for i in range(1,n+1):
        fact*=i
    return fact

factorial(5)
```

```python
#factorial formula
#n! = n * ( n - 1 ) *...*1

# This code defines a function called "factorial" which takes an input "n". The function uses a for loop to iterate through the range of numbers 
# from 1 to n+1. For each number in that range, it multiplies the current value of x by the number in the range. At the end of the loop, 
# the function returns the final value of x, which will be the factorial of the input number "n".

# The Factorial of a positive integer n is the product of all positive integers less than or equal to n. 
# For example, the factorial of 6 (written "6!") is 6 * 5 * 4 * 3 * 2 * 1 = 720.

# So this function takes an input of any positive integer, and returns the factorial of that number.
```

```python
list1 = [4,4,7,6,2,2,3,3,3,3,9]
def unique(lst_un):
    uni_lst1 = []
    for i in lst_un:
        if i not in uni_lst1:
            uni_lst1.append(i)
    return uni_lst1

unique(list1)
```

## 8. Write a function that returns the mode of a list, i.e.: the element that appears the most times.
`NOTE: You should not use count... 🧐`

```python
def mode_counter(arr):
    counter_dict = {}
    
    for i in arr:
        if i in counter_dict:
            counter_dict[i] += 1
        else:
            counter_dict[i] = 1
    
    mode = None
    max_count = 0
    for key, value in counter_dict.items():
        if value > max_count:
            max_count = value
            mode = key
    
    return mode

# Example usage
list1 = [1, 2, 3, 2, 4, 2, 5, 3, 3]
mode_counter(list1)
```

```python
# This will test your function 
test_mode(mode_counter)
```

## 9. Write a function that calculates the standard deviation of a list.
`NOTE: Do not use any libraries or already built functions. 😉`

```python
import math
def Standard_Dev (lst):
    sum = 0
    for i in lst:
        sum += i
    N = len(lst)
    mean = sum/N
    #print(mean)
    var = 0
    for j in lst:
        var += (j-mean)**2
    varience = var/(N-1)
    SD = math.sqrt(varience)
    return SD

list1 = [2, 2, 40, 8, 11]
Standard_Dev(list1)
```

```python
# This will test your function 
test_stdev(Standard_Dev)
```

## 11. Write a function that receives a string of comma separated words and returns a string of comma separated words sorted alphabetically.

`NOTE: You may use sorted but not split and definitely no join! 🤪`

```python
lst1 = "hello, how, are, you, doing"

def sort_alpha (strr1):
    clean_list = [word.strip() for word in strr1.split(',')]
    #print(clean_list)

    x = sorted(clean_list)
    final_string = ', '.join(x)
    return final_string

sort_alpha(lst1)
```

## 12. Write a function to check if a given password is strong (at least 8 characters, at least one lower case, at least one upper case, at least one number and at least one special character). It should output True if strong and False if not.

```python
def check_pass(password):
    special_chars = "!@#$%^&*()_+-={}[]|\\:;\"'<>,.?/~`"
    if len(password) <8:
        #print("Length of the password should be at least 8 characters")
        return False
    elif not any(char.isdigit() for char in password):  # Check for at least one digit
        return False
    elif any(char.islower() for char in password) == False:
        #print("False - Password should have at least one lower case")
        return False
    elif any(char.isupper() for char in password) == False:
        #print("False - Password should have at least one upper case")
        return False
    elif any(char in special_chars for char in password) == False:
        #print("False - Password should have at least one special character")
        return False
    else:
        return True

password1 = "&nukBcAkox1{"
check_pass(password1)
```

```python
# This will test your function 
test_pass(check_pass)
```
